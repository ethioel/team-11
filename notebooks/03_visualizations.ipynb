```python
# Verify figures exist
import os
figures = sorted([f for f in os.listdir(FIGURES_DIR) if f.startswith('fig') and f.endswith('.png')])
print(f"\nGenerated figures ({len(figures)}):")
for fig in figures:
    filepath = FIGURES_DIR / fig
    size_mb = os.path.getsize(filepath) / (1024 * 1024)
    print(f"  ✓ {fig} ({size_mb:.2f} MB)")

# Display figure captions
captions_file = FIGURES_DIR / 'figure_captions.md'
if captions_file.exists():
    with open(captions_file, 'r') as f:
        print(f"\nFigure captions file exists with {len(f.readlines())} lines")
```
